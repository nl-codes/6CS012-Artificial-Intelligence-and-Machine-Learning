```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Path to your subset directory
data_path = "./Dataset/Hotel_Reviews.csv"

# Load your dataset
df = pd.read_csv(data_path)

# 1. Class Distribution Analysis
plt.figure(figsize=(10, 6))
sns.countplot(x="Rating", data=df, hue="Rating", palette="viridis", legend=False)

plt.title("Class Distribution: Frequency of Hotel Ratings")
plt.xlabel("Rating (1-5)")
plt.ylabel("Number of Reviews")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()

# Print exact counts for numerical analysis
print("Rating Counts:\n", df["Rating"].value_counts().sort_index())
```

```python
# 2. Sample Visualization of Reviews
print("-" * 30)
print("SAMPLE REVIEWS PER RATING")
print("-" * 30)

# Iterate through each rating class (1 to 5)
for rating in range(1, 6):
    print(f"\n--- Rating {rating} ---")

    # Get one random sample for the current rating
    sample = df[df["Rating"] == rating]["Review"].iloc[0]

    # Print the first 300 characters to keep it concise
    print(f"Sample: {sample[:300]}...")
```
