### Task - 1:

Create a Python program that converts between different units of measurement.

• The program should:

1. Prompt the user to choose the type of conversion (e.g., length, weight, volume).
2. Ask the user to input the value to be converted.
3. Perform the conversion and display the result.
4. Handle potential errors, such as invalid input or unsupported conversion types.

• Requirements:

1. Functions: Define at least one function to perform the conversion.
2. Error Handling: Use try-except blocks to handle invalid input (e.g., non-numeric values).
3. User Input: Prompt the user to select the conversion type and input the value.
4. Docstrings: Include a docstring in your function to describe its purpose, parameters, and
   return value.

• Conversion Options:

1. Length:
    - Convert meters (m) to feet (ft).
    - Convert feet (ft) to meters (m).
2. Weight:
    - Convert kilograms (kg) to pounds (lbs).
    - Convert pounds (lbs) to kilograms (kg).
3. Volume:
    - Convert liters (L) to gallons (gal).
    - Convert gallons (gal) to liters (L).


```python
# Conversion Registry: Map of conversion types
# [Factor, Metric Unit, Imperial Unit]
# The factor converts Metric to Imperial (Metric * Factor = Imperial)

registry = {
    "length": [3.28084, "m", "ft"],
    "weight": [2.20462, "kg", "lbs"],
    "volume": [0.264172, "L", "gal"],
}


def handle_conversion(to_imperial, conversion_type):
    """
    Performs the conversion for length, weight, or volume based on a boolean flag.

    Parameters:
        to_imperial (bool): If True, multiplies by factor. If False, divides.
        conversion_type (str): The category ('length', 'weight', or 'volume').

    Returns:
        float: The converted value, or None if an error occurs.
    """

    factor, metric_unit, imperial_unit = registry[conversion_type]

    from_unit = metric_unit if to_imperial else imperial_unit

    try:
        val = float(input(f"Enter {from_unit} value: "))

        if to_imperial:
            result = val * factor
            print(f"{val} {metric_unit} is {result:.2f} {imperial_unit}")
        else:
            result = val / factor
            print(f"{val} {imperial_unit} is {result:.2f} {metric_unit}")

        return result
    except ValueError:
        print("Error: Please enter a valid numerical value.")
        return None


def units_converter():
    """
    Main controller to prompt user for category and direction.

    Returns:
        float: The result of the conversion, or None for invalid choices.
    """
    print("--- Professional Multi-Unit Converter ---")
    print("1. Length (m/ft)\n2. Weight (kg/lbs)\n3. Volume (L/gal)")
    choice = input("Select category (1-3): ")

    # Map input choice to the string key used in handle_conversion
    type_map = {"1": "length", "2": "weight", "3": "volume"}

    if choice not in type_map:
        print("Invalid category.")
        return None

    # Get the specific units for this category from the registry
    category = type_map[choice]
    _, metric, imperial = registry[category]

    # Dynamically show the units in the direction prompt
    print(f"\nSelect Direction for {category.capitalize()}:")
    print(f"1. Metric ({metric}) to Imperial ({imperial})")
    print(f"2. Imperial ({imperial}) to Metric ({metric})")

    direction = input("Select direction (1 or 2): ")

    if direction not in ["1", "2"]:
        print("Invalid direction.")
        return None

    # Determine boolean: True if Metric -> Imperial (1), False otherwise
    is_imperial = direction == "1"
    # Call the single handler with the determined parameters
    handle_conversion(is_imperial, type_map[choice])


units_converter()
```

### Task - 2:

Create a Python program that performs various mathematical operations on a list of numbers.

• The Program should:

1. Prompt the user to choose an operation (e.g., find the sum, average, maximum, or minimum
   of the numbers).
2. Ask the user to input a list of numbers (separated by spaces).

3. Perform the selected operation and display the result.
4. Handle potential errors, such as invalid input or empty lists.

• Requirements:

1. **Functions:** Define at least one function for each operation (sum, average, maximum, mini-
   mum).

2. **Error Handling:** Use try-except blocks to handle invalid input (e.g., non-numeric values or
   empty lists).
3. **User Input:** Prompt the user to select the operation and input the list of numbers.
4. **Docstrings:** Include a docstring in each function to describe its purpose, parameters, and
   return value.


```python
def find_sum(numbers):
    """
    Calculates the total sum of a list of numbers

    Parameters: numbers (list of float)
    Returns: float
    """
    total = 0.0
    for num in numbers:
        total += num
    return total


def find_average(numbers):
    """
    Calculates the average of a list of numbers.

    Parameters: numbers (list of float)
    Returns: float
    """
    if not numbers:
        return 0.0
    total = find_sum(numbers)
    count = 0
    for _ in numbers:
        count += 1
    return total / count


def find_max(numbers):
    """
    Finds the largest value in the list

    Parameters: numbers (list of float)
    Returns: float
    """
    if not numbers:
        return None
    maximum = numbers[0]
    for num in numbers:
        if num > maximum:
            maximum = num
    return maximum


def find_min(numbers):
    """
    Finds the smallest value in a list

    Parameters: numbers (list of float)
    Returns: float
    """
    if not numbers:
        return None
    minimum = numbers[0]
    for num in numbers:
        if num < minimum:
            minimum = num
    return minimum


def math_operations_tool():
    """
    Main controller for the Math Operations program.
    Prompts user for list of numbers and an operation, then using helper funcitons operates the given list and displays the result.
    """
    print("--- Mathematical Operations on List ---")

    try:
        # Get user input
        raw_input = input("Enter a list of numbers separated by spaces: ")

        numbers = []
        for x in raw_input.split():
            numbers.append(float(x))

        if not numbers:
            print("Error: The list is empty.")
            return

        print("\nChoose an operation:")
        print("1. Sum\n2. Average\n3. Maximum\n4. Minimum")
        choice = input("Enter choice (1-4): ")

        if choice == "1":
            print(f"Result (Sum): {find_sum(numbers)}")
        elif choice == "2":
            print(f"Result (Average): {find_average(numbers)}")
        elif choice == "3":
            print(f"Result (Maximum): {find_max(numbers)}")
        elif choice == "4":
            print(f"Result (Minimum): {find_min(numbers)}")
        else:
            print("Invalid operation choice.")

    except ValueError:
        print("Error: Invalid input. Please enter only numbers separated by spaces.")
    except Exception as e:
        print(f"An error occurred: {e}")


math_operations_tool()
```

```python
def flatten(lst):
    """
    Flattens a nested list into a single list

    Parameters:
        lst (list[list]): A nested list

    Returns:
        flat_list (list[int]): A flattened list containing every element of the nested list
    """
    flat_list = []
    # for sub_list in lst:
    #     for ele in sub_list:
    #         flat_list.append(ele)

    for sub_list in lst:
        flat_list.extend(sub_list)

    return flat_list


nested = [[1, 2], [3, 4], [5]]
print(flatten(nested))
```

### 3. Sum of All Elements in a Nested List:

Write a Python function that calculates the sum of all the numbers in a nested list (regardless of depth).

* **Requirements**:

    -  Define a function `sum_nested(lst)` that takes a nested list `lst` and returns the sum of all
the elements.

    - Example: For the input [[1, 2], [3, [4, 5]], 6], the output should be 21.

```python
def sum_nested(lst):
    """
    Calculates the sum of all numbers in a nested list
    To account for any depth of list, recursive function is used

    Parameters:
        lst (list[list]): Nested list of numbers

    Returns:
        total (int): Sum of all numbers
    """
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)
        else:
            total += item
    return total


print(sum_nested([[1, 2], [3, [4, 5]], 6]))
```

### 4. Remove Specific Element from a Nested List:
Write a Python function that removes all occurrences of a specific element from a nested list.

* **Requirements**:

    - Define a function `remove_element(lst, elem)` that removes `elem` from `lst` and returns the
modified list.

    - Example: For the input lst = [[1, 2], [3, 2], [4, 5]] and elem = 2, the output should
be [[1], [3], [4, 5]].

```python
def remove_element(lst, elem):
    """
    Removes all occurrences of an element from a nested list

    Parameters:
        lst (list[list]): Nested list
        elem (int): Element to remove

    Returns:
        result (list): Modified nested list
    """
    result = []
    for item in lst:
        if isinstance(item, list):
            sub = remove_element(item, elem)
            result.append(sub)
        elif item != elem:
            result.append(item)
    return result


print(remove_element([[1, 2], [3, 2], [4, 5]], 2))
```

### 5. Find the Maximum Element in a Nested List:
Write a Python function that finds the maximum element in a nested list (regardless of depth).

* **Requirements**:

    - Define a function `find_max(lst)` that takes a nested list `lst` and returns the maximum
element.
    - Example: For the input [[1, 2], [3, [4, 5]], 6], the output should be 6.

```python
def find_max(lst):
    """
    Finds the maximum element in a nested list

    Parameters:
        lst (list[list]): Nested list

    Returns:
        max_val (int): Maximum value in lst
    """
    max_val = float("-inf")

    for item in lst:
        if isinstance(item, list):
            sub_max = find_max(item)
            if sub_max > max_val:
                max_val = sub_max
        else:
            if item > max_val:
                max_val = item
    return max_val


print(find_max([[1, 2], [3, [4, 5]], 6]))
```

### 6. Count Occurrences of an Element in a Nested List:
Write a Python function that counts how many times a specific element appears in a nested list.

* **Requirements**:

    - Define a function `count_occurrences(lst, elem)` that counts the occurrences of `elem` in
the nested list `lst`.

    - Example: For the input lst = [[1, 2], [2, 3], [2, 4]] and elem = 2, the output should
be 3.

```python
def count_occurrences(lst, elem):
    """
    Counts occurence of an element in a nested list

    Parameters:
        lst (list[list]): Nested list
        elem (int): Element to count

    Returns:
        count (int): Number of occurrences
    """
    count = 0
    for item in lst:
        if isinstance(item, list):
            count += count_occurrences(item, elem)
        elif item == elem:
            count += 1
    return count


print(count_occurrences([[1, 2], [2, 3], [2, 4]], 2))
```

### 7. Flatten a List of Lists of Lists:
Write a Python function that flattens a list of lists of lists into a single list, regardless of the depth.
* **Requirements**:

    - Define a function `deep_flatten(lst)` that takes a deeply nested list `lst` and returns a single
flattened list.

    - Example: For the input [[[1, 2], [3, 4]], [[5, 6], [7, 8]]], the output should be
[1, 2, 3, 4, 5, 6, 7, 8].

```python
def deep_flatten(lst):
    """
    Flattens a multi-layered nested list into a single list

    Parameters:
        lst (list[list]): Nested list

    Returns:
        result (list[int]): Flattened list
    """
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(deep_flatten(item))
        else:
            result.append(item)
    return result


print(deep_flatten([[[1, 2], [3, 4]], [[5, 6], [7, 8]]]))
```

### 8. Nested List Average:
Write a Python function that calculates the average of all elements in a nested list.

* **Requirements**:

    - Define a function `average_nested(lst)` that takes a nested list `lst` and returns the average
of all the elements.
    
    - Example: For the input [[1, 2], [3, 4], [5, 6]], the output should be 3.5.

```python
def recursive_helper(sublist):
    """
    Recursively traverses a nested list to calculate the sum and count of all elements

    Args:
        sublist (list): A list that may contain integers, floats, or other lists

    Returns:
        tuple: A tuple containing (total_sum, element_count).
    """
    total_sum = 0
    element_count = 0
    for item in sublist:
        if isinstance(item, list):
            t, c = recursive_helper(item)
            total_sum += t
            element_count += c
        else:
            total_sum += item
            element_count += 1
    return total_sum, element_count


def average_nested(lst):
    """
    Calculates the average of all elements in a nested list.

    Parameters:
        lst (list): Nested list containing numerical values.

    Returns:
        float: Average of all elements, or 0 if the list is empty.
    """
    # get total sum and total number of elements from recursive function
    list_sum, ele_count = recursive_helper(lst)
    return list_sum / ele_count if ele_count != 0 else 0


print(average_nested([[1, 2], [3, 4], [5, 6]]))
```

### Problem - 4: Matrix Operations:

For the following arrays:

`A = np.array([[3,4],[7,8]])`

`B = np.array([[5,3],[2,1]]);`


```python
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

print("A : \n", A)
print("B : \n", B)
```

#### Prove following with Numpy:

  1. Prove A.A<sup>−1</sup> = I.


```python
A_inv = np.linalg.inv(A)
print("Inverse of matrix A : \n", A_inv, end="\n\n")

dot_A_Ainv = np.dot(A, A_inv)
print("Dot product of A and A inverse : \n", dot_A_Ainv, end="\n\n")

check = np.allclose(dot_A_Ainv, np.eye(A.shape[0]))
print(f"Is it effectively the Identity Matrix? {check}")
```


  2. Prove AB &ne; BA.


```python
dot_A_B = np.dot(A, B)
dot_B_A = np.dot(B, A)

print("Dot product of A and B (A.B) : \n", dot_A_B, end="\n\n")
print("Dot product of B and A (B.A) : \n", dot_B_A, end="\n\n")

print(
    f"Is each element of A.B close to B.A? \n{np.isclose(dot_A_B, dot_B_A)}", end="\n\n"
)

is_commutative = np.allclose(dot_A_B, dot_B_A)
print(f"Is A.B equal to B.A? {is_commutative}")
```


  3. Prove (AB)<sup>T</sup> = B<sup>T</sup>A<sup>T</sup>.

```python
dot_A_B_trans = np.transpose(dot_A_B)
B_trans = np.transpose(B)
A_trans = np.transpose(A)
dot_Btrans_Atrans = np.dot(B_trans, A_trans)

print("Transpose of dot product of A and B (A.B)T : \n", dot_A_B_trans, end="\n\n")
print(
    "Dot product of B transpose and A transpose (B)T.(A)T : \n",
    dot_Btrans_Atrans,
    end="\n\n",
)

comparison = np.isclose(dot_A_B_trans, dot_Btrans_Atrans)
print(f"Is (A.B)T equal to BT.AT? \n{comparison}")

print(f"\nVerification: {np.all(comparison)}")
```

## 10.2 Experiment: How Fast is Numpy?

In this exercise, you will compare the performance and implementation of operations using plain Python
lists (arrays) and NumPy arrays. Follow the instructions:

```python
def time_the_operation(func, *args):
    """
    Measures the time taken by a function to execute.

    Params:
        func (callable): The function to time.
        *args: Arguments to pass to the function.

    Returns:
        float: Time taken in seconds for the function to execute.
    """
    start_time = time.time()
    func(*args)
    end_time = time.time()
    return end_time - start_time


# Initialize data
SIZE = 1_000_000
list_a = list(range(SIZE))
list_b = list(range(SIZE))
arr_a = np.arange(SIZE)
arr_b = np.arange(SIZE)
```

### 1. Element-wise Addition:

• Using Python Lists, perform element-wise addition of two lists of size 1, 000, 000. Measure and Print the time taken for this operation.

• Using Numpy Arrays, Repeat the calculation and measure and print the time taken for this operation.

```python
def list_addition(l1, l2):
    """
    Performs element-wise addition of two lists.

    Params:
        l1 (list): First list of numbers.
        l2 (list): Second list of numbers.
    """
    for x, y in zip(l1, l2):
        x + y


def numpy_addition(a1, a2):
    """
    Performs element-wise addition of two numpy arrays.

    Params:
        a1 (np.ndarray): First array.
        a2 (np.ndarray): Second array.
    """
    a1 + a2


t_add_list = time_the_operation(list_addition, list_a, list_b)
t_add_np = time_the_operation(numpy_addition, arr_a, arr_b)

print(f"--- Element-wise Addition (Size: {SIZE:,}) ---")
print(f"Python List Time: {t_add_list:.6f} seconds")
print(f"NumPy Array Time: {t_add_np:.6f} seconds")
print(f"Speedup: {t_add_list / t_add_np:.2f}x")
```

### 2. Element-wise Multiplication

• Using Python Lists, perform element-wise multiplication of two lists of size 1, 000, 000. Measure and Print the time taken for this operation.

• Using Numpy Arrays, Repeat the calculation and measure and print the time taken for this operation.


```python
def list_multiplication(l1, l2):
    """
    Performs element-wise multiplication of two lists.

    Params:
        l1 (list): First list of numbers.
        l2 (list): Second list of numbers.

    Returns:
        list: List containing element-wise products.
    """
    for x, y in zip(l1, l2):
        x * y


def numpy_multiplication(a1, a2):
    """
    Performs element-wise multiplication of two numpy arrays.

    Params:
        a1 (np.ndarray): First array.
        a2 (np.ndarray): Second array.
    """
    a1 * a2


t_mul_list = time_the_operation(list_multiplication, list_a, list_b)
t_mul_np = time_the_operation(numpy_multiplication, arr_a, arr_b)

print(f"--- Element-wise Multiplication (Size: {SIZE:,}) ---")
print(f"Python List Time: {t_mul_list:.6f} seconds")
print(f"NumPy Array Time: {t_mul_np:.6f} seconds")
print(f"Speedup: {t_mul_list / t_mul_np:.2f}x")
```

### 3. Dot Product

• Using Python Lists, compute the dot product of two lists of size 1, 000, 000. Measure and Print the time taken for this operation.

• Using Numpy Arrays, Repeat the calculation and measure and print the time taken for this operation.

```python
def list_dot_product(l1, l2):
    """
    Computes the dot product of two lists.

    Params:
        l1 (list): First list of numbers.
        l2 (list): Second list of numbers.
    """
    dot = 0
    for x, y in zip(l1, l2):
        dot += x * y


def numpy_dot_product(a1, a2):
    """
    Computes the dot product of two numpy arrays.

    Params:
        a1 (np.ndarray): First array.
        a2 (np.ndarray): Second array.
    """
    np.dot(a1, a2)


t_dot_list = time_the_operation(list_dot_product, list_a, list_b)
t_dot_np = time_the_operation(numpy_dot_product, arr_a, arr_b)

print(f"--- Dot Product (Size: {SIZE:,}) ---")
print(f"Python List Time: {t_dot_list:.6f} seconds")
print(f"NumPy Array Time: {t_dot_np:.6f} seconds")
print(f"Speedup: {t_dot_list / t_dot_np:.2f}x")
```

### 4. Matrix Multiplication

• Using Python lists, perform matrix multiplication of two matrices of size 1000x1000. Measure and print the time taken for this operation.

• Using NumPy arrays, perform matrix multiplication of two matrices of size 1000x1000. Measure and print the time taken for this operation.

```python
MAT_SIZE = 1000
mat_list_a = [[float(i) for i in range(MAT_SIZE)] for _ in range(MAT_SIZE)]
mat_list_b = [[float(i) for i in range(MAT_SIZE)] for _ in range(MAT_SIZE)]
mat_np_a = np.array(mat_list_a)
mat_np_b = np.array(mat_list_b)


def list_matrix_mul(A, B):
    """
    Performs matrix multiplication using Python lists.

    Params:
        A (list[list[float]]): First matrix as a list of lists.
        B (list[list[float]]): Second matrix as a list of lists.
    """
    # Initialize result matrix with zeros
    result = [[0.0 for _ in range(len(B[0]))] for _ in range(len(A))]
    for i in range(len(A)):
        for j in range(len(B[0])):
            for k in range(len(B)):
                result[i][j] += A[i][k] * B[k][j]


def numpy_matrix_mul(A, B):
    """
    Performs matrix multiplication using numpy arrays.

    Params:
        A (np.ndarray): First matrix.
        B (np.ndarray): Second matrix.
    """
    np.matmul(A, B)


print(f"--- Matrix Multiplication ({MAT_SIZE}x{MAT_SIZE}) ---")
# NumPy Timing
t_mat_np = time_the_operation(numpy_matrix_mul, mat_np_a, mat_np_b)
print(f"NumPy Array Time: {t_mat_np:.6f} seconds")

# List Timing (Slow!)
print("Calculating List Matrix Mul")
t_mat_list = time_the_operation(list_matrix_mul, mat_list_a, mat_list_b)
print(f"Python List Time: {t_mat_list:.6f} seconds")

print(f"Speedup: {t_mat_list / t_mat_np:.2f}x")
```
