The reason we only we gray image is because slicing a single channel, such as lena_arr[:, :, 0], converts a 3D color array into a 2D intensity matrix. When the 2D grid is passed to Image.fromarray(), the library treats it as a grayscale image.

At this stage, the computer no longer "knows" these numbers represent red, green, or blue. It only sees raw brightness values where 0 is black and 255 is white. Without the other two color dimensions to provide context, the result is a flat, gray scaled image.

**Pseudo-coloring**

To see the channels in their respective colors, Matplotlib's pyplot and its colormap (cmap) feature can be used. Instead of relying on the image's internal color data, we will use plt.imshow() that finds a lookup table to maps those 2D intensity values to a specific color gradient, such as "Reds" or "Greens" or "Blues".



```python
# Visualize the three color channels of the Lena image using matplotlib
red_channel = lena_arr[:, :, 0]
green_channel = lena_arr[:, :, 1]
blue_channel = lena_arr[:, :, 2]

plt.figure(figsize=(15, 5))

# Red channel visualization
plt.subplot(1, 3, 1)
plt.title("Red Channel")
plt.imshow(red_channel, cmap="Reds")
plt.axis("off")

# Green channel visualization
plt.subplot(1, 3, 2)
plt.title("Green Channel")
plt.imshow(green_channel, cmap="Greens")
plt.axis("off")

# Blue channel visualization
plt.subplot(1, 3, 3)
plt.title("Blue Channel")
plt.imshow(blue_channel, cmap="Blues")
plt.axis("off")

plt.show()
```

# 3 Image Compression and Decompression using PCA.

In this exercise, build a PCA from scratch using explained variance method for image compression task.
You are expected to compute the necessary matrices from the scratch.

Dataset: Use image of your choice.


1. Load and Prepare Data:

    * Fetch an image of you choice.{If colour convert to grayscale}

    * Center the dataset - Standaridze the Data.

    * Calculate the covaraince matrix of the Standaridze data.



```python
# Load and display a colored image (Goku)
colored_image = Image.open("goku.png")
display(colored_image)
```

```python
# Convert the colored image to grayscale
image = Image.open("goku.png").convert("L")
display(image)
```

```python
# Convert image to numpy array
image_array = np.array(image)

# Get image dimensions
height, width = image_array.shape
print("Shape of image : ", height, width)

# Reshape the image into a 2D array where each row represents a row of pixels
# This allows PCA to process each row as an observation and find patterns across pixel intensities
data = image_array.reshape(height, width)
```

```python
# Standardizing the data

# Compute the mean of each column (feature)
mean = np.mean(data, axis=0)

# Subtract mean to center the data
centered_data = data - mean

# Calculate covariance of the centered data
cov_matrix = np.cov(centered_data, rowvar=False)
```

2. Eigen Decomposition and Identifying Principal Components:

    * Compute Eigen Values and Eigen Vectors.

    * Sort the eigenvalues in descending order and choose the top k eigenvectors corresponding to the highest eigenvalues.

    * Identify the Principal Components with the help of cumulative Sum plot.


```python
# Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# Sort eigenvalues and corresponding eigenvectors in descending order
sorted_indices = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[sorted_indices]
eigenvectors = eigenvectors[:, sorted_indices]

# Plot the explained variance ratio
explained_variance_ratio = eigenvalues / np.sum(eigenvalues)
cumulative_variance = np.cumsum(explained_variance_ratio)
plt.plot(cumulative_variance)
plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.grid(True)
plt.show()
```

```python
k = 215
components = eigenvectors[:, :k]
```

3. Reconstruction and Experiment:

    * Reconstruction: Transform the original data by multiplying it with the selected eigenvectors(PCs) to obtain a lower-dimensional representation.

    * Experiments: Pick Four different combination of principal components with various explained variance value and compare the result.

    * Display the Results and Evaluate.


```python
# Project the data onto the principal components
compressed_data = np.dot(centered_data, components)
```

```python
# Reconstruct the image from compressed data
decompressed_data = np.dot(compressed_data, components.T) + mean
```

```python
print("original shape : ", image_array.shape)
print("compressed shape : ", compressed_data.shape)
print("decompressed shape : ", decompressed_data.shape)
```

```python
# Visualize the original, compressed, and decompressed images using PCA components
plt.figure(figsize=(12, 6))

# Original Image
plt.subplot(1, 3, 1)
plt.imshow(image_array, cmap="gray")
plt.title("Original Image")
plt.axis("off")

# Compressed Representation
plt.subplot(1, 3, 2)
plt.imshow(compressed_data, cmap="gray", aspect="auto")
plt.title(f"Compressed Image (k={k})")
plt.axis("off")

# Decompressed Image
plt.subplot(1, 3, 3)
plt.imshow(decompressed_data, cmap="gray")
plt.title("Decompressed Image")
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Plot only the first 200 components
plt.figure(figsize=(10, 6))
plt.plot(
    range(1, 201), cumulative_variance[:200], marker="o", markersize=2, linestyle="-"
)

plt.title("Cumulative Explained Variance (First 200 Components)")
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance Ratio")
plt.xlim(0, 200)
plt.ylim(0, 1.05)
plt.grid(True, linestyle="--", alpha=0.7)
plt.show()

variance_at_100 = cumulative_variance[199] * 100
print(f"\nTotal variance explained by the first 200 components: {variance_at_100:.2f}%")
```

```python
# Experiment: Reconstruct the image using different numbers of PCA components
k_values = [image_array.shape[1], 1, 5, 25, 50, 75, 100, 150, 200]
total_variance = np.sum(eigenvalues)

# Setup a 3x3 grid for the 9 images
plt.figure(figsize=(18, 18))

for i, k in enumerate(k_values):
    # 1. Select the top k components
    components_subset = eigenvectors[:, :k]

    # 2. Calculate Explained Variance for this k
    current_variance = (np.sum(eigenvalues[:k]) / total_variance) * 100

    # 3. Project and Reconstruct
    compressed = np.dot(centered_data, components_subset)
    reconstructed = np.dot(compressed, components_subset.T) + mean

    # 4. Visualization
    plt.subplot(3, 3, i + 1)  # 3 rows, 3 columns
    plt.imshow(reconstructed, cmap="gray")

    # Title
    title_label = "Original (Full)" if i == 0 else f"k = {k}"
    plt.title(f"{title_label}\nExplained Var: {current_variance:.2f}%", fontsize=12)
    plt.axis("off")

plt.tight_layout()
plt.subplots_adjust(top=0.92)
plt.suptitle("PCA Image Reconstruction Experiment", fontsize=20)
plt.show()
```
